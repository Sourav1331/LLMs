# 12. Deployable LLM Assistant

This notebook is the final step of the project. The earlier notebooks explain how LLM components work internally. This notebook connects the saved fine-tuned checkpoint to a deployable API.

## Goal

By the end of this step, the project can:

- load the fine-tuned model from `fine_tuned_distilgpt2_instruction_demo/`
- accept a user prompt
- generate a response
- expose generation through a FastAPI service
- be prepared for local or cloud deployment

## Start the API

Run this from the project root in a terminal:

```bash
uvicorn app.api:app --reload --host 0.0.0.0 --port 8000
```

Then open:

```text
http://localhost:8000/docs
```

````python
import requests

url = "http://localhost:8000/health"

try:
    response = requests.get(url, timeout=10)
    print(response.json())
except Exception as exc:
    print("API is not running yet. Start it with: uvicorn app.api:app --reload --host 0.0.0.0 --port 8000")
    print(exc)
````

````python
import requests

payload = {
    "prompt": "Explain tokenization in simple words.",
    "max_new_tokens": 80,
    "temperature": 0.7,
    "top_p": 0.9,
    "do_sample": True,
}

try:
    response = requests.post("http://localhost:8000/generate", json=payload, timeout=60)
    print(response.json())
except Exception as exc:
    print("Generation request failed. Make sure the API server is running.")
    print(exc)
````
